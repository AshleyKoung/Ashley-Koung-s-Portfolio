從兩個table中查詢特定的欄位，並且**移除重複**的內容：

基本上我們會從SELECT依序開使撰寫查詢語句，最後為了消除重複的內容，再往前去新增關鍵**Distinct**

```python
SELECT DISTINCT sname, major
FROM student, Apply
WHERE student.sID = Apply.sID;
```

**ORDER BY** 排序
告訴資料庫你希望output依照哪個column進行排序

假設你輸入 **ORDER BY** GPA **desc**;
則會根據GPA column進行降序排列，也就是表格由GPA高排到低
反之**asc**則為升序。

```python
SELECT student.sID, sname, GPA
FROM student, College, Apply
WHERE student.sID = Apply.sID and Apply.cName = College.cName
ORDER BY GPA desc;
```

從table中**尋找特定規格的內容**：

假設輸入「％bio」則表示想查詢所有內容結尾有bio的內容，假設輸入「bio%」則表示想查詢所有開頭為bio的內容。

以下範例說明想從Apply table中查詢sID和major欄位，並且搜尋條件為major的內容中不論開頭或結尾有出現bio的。

```python
SELECT sID,major
FROM Apply
WHERE major like '%bio%';
```

如果想要將SELECT呈現的欄位改名字，則可以在欄位名後方運用 **AS** 進行更名。

以下範例，我們想將major欄位在output中更名study_major:

```python
SELECT
sID,
major AS study_major\
FROM Apply
WHERE major like '%bio%';
```
