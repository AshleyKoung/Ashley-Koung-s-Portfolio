從兩個table中查詢特定的欄位，並且**移除重複**的內容：

基本上我們會從SELECT依序開使撰寫查詢語句，最後為了消除重複的內容，再往前去新增關鍵**Distinct**

```python
SELECT DISTINCT sname, major
FROM student, Apply
WHERE student.sID = Apply.sID;
```

**ORDER BY** 排序
告訴資料庫你希望output依照哪個column進行排序

假設你輸入 **ORDER BY** GPA **desc**;
則會根據GPA column進行降序排列，也就是表格由GPA高排到低
反之**asc**則為升序。

```python
SELECT student.sID, sname, GPA
FROM student, College, Apply
WHERE student.sID = Apply.sID and Apply.cName = College.cName
ORDER BY GPA desc;
```

從table中**尋找特定規格的內容**：

假設輸入「％bio」則表示想查詢所有內容結尾有bio的內容，假設輸入「bio%」則表示想查詢所有開頭為bio的內容。

以下範例說明想從Apply table中查詢sID和major欄位，並且搜尋條件為major的內容中不論開頭或結尾有出現bio的。

```python
SELECT sID,major
FROM Apply
WHERE major like '%bio%';
```

如果想要將SELECT呈現的欄位改名字，則可以在欄位名後方運用 **AS** 進行更名。

以下範例，我們想將major欄位在output中更名study_major:

```python
SELECT
sID,
major AS study_major\
FROM Apply
WHERE major like '%bio%';
```

**UNION聯集** 可以將兩個表格 **垂直合併**

舉例來說，如果兩個表格都有name，那麼你可以將兩個表格中的name合併起來在新的表格中呈現。
通常來說當使用UNION時，系統會自動將重複的內容移除，但我們不保證每個系統都能做到。
假設你希望呈現所有重複的內容，則需使用 **UNION ALL**


以下範例作為示範：


```python
SELECT cName AS name FROM College
UNION
SELECT sName AS name FROM Student;
```

**INTERSECT交集**
運用方式跟AND相同，用於查詢一個結果集內同時亦存在於第二個查詢結果的集內。

但，並不是每個資料庫管理系統都支援INTERSECT。
假如管理系統不支援，則需要傳統的方式查詢，以下舉例兩種不同的查詢方式：

**INTERSECT**

```python
SELECT sID FROM Apply WHERE major = 'CS'
INTERSECT
SELECT sID FROM Apply WHERE major = 'EE';
```

**沒有INTERSECT的查詢方式**

```python
SELECT DISTINCT A1.sID
FROM Apply A1, Apply A2
WHERE A1.sID = A2.sID AND A1.major = 'CS' AND A2.major = 'EE';
```

**EXCEPT排除** 運用於查詢一個結果存在於第一個結果集內但排除存在於第二個結果集。

但，並不是每個資料庫管理系統都支援EXCEPT。 假如管理系統不支援，則需要傳統的方式查詢，以下舉例兩種不同的查詢方式

**EXCEPT**

```python
SELECT sID FROM Apply WHERE major = 'CS'
EXCEPT
SELECT sID FROM Apply WHERE major = 'EE';
```

**沒有EXCEPT的查詢方式**

```python
SELECT DISTINCT A1.sID
FROM Apply A1, Apply A2
WHERE A1.sID =A2.sID AND A1.major = 'cs' and A2.major <> 'EE';
```

使用**EXISTS**來找出最大與最小值

範例如下：

1.   對於每個來自 C1 的學生（主查詢中的學生），子查詢會檢查 C2 表中是否存在一個學生的 GPA 比當前學生的 GPA 更高（C2.GPA > C1.GPA）。
2.   如果子查詢找不到任何比當前學生（C1）的 GPA 更高的記錄，表示這個學生在 C1 中是成績最好的之一。
因此，NOT EXISTS 條件為 TRUE，這個學生就會被返回到結果中。

```python
SELECT sName, GPA
FROM Student C1
WHERE NOT EXISTS (select * from Student C2 where C2.GPA > C1.GPA);
```
